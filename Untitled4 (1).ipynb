### Floyd-Warshall Algorithm Implementation
The Floyd-Warshall algorithm is a dynamic programming algorithm used to find the shortest paths between all pairs of vertices in a weighted graph (with positive or negative edge weights, but no negative cycles).

It works by incrementally allowing intermediate vertices in the path: For each pair of vertices $(u, v)$, it checks if a path through an intermediate vertex $k$ is shorter than the currently known path between $u$ and $v$.

```python
import math

def floyd_warshall(vertices, graph_matrix):
    """
    Computes the shortest paths between all pairs of vertices.
    :param vertices: List of vertices
    :param graph_matrix: 2D dictionary representing adjacency matrix weight,
                         where graph_matrix[u][v] is the weight of edge u -> v.
    :return: A 2D dictionary of shortest distances between all pairs.
    """
    # Initialize distance table with infinity
    dist = {u: {v: math.inf for v in vertices} for u in vertices}

    # Distance from a node to itself is 0
    for v in vertices:
        dist[v][v] = 0

    # Set initial edge weights
    for u in graph_matrix:
        for v in graph_matrix[u]:
            dist[u][v] = graph_matrix[u][v]

    # Floyd-Warshall main loop
    for k in vertices:
        for i in vertices:
            for j in vertices:
                if dist[i][j] > dist[i][k] + dist[k][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]

    return dist

# --- Example Usage ---
fw_vertices = ['1', '2', '3', '4']
# Adjacency representation using dictionary
fw_graph = {
    '1': {'2': 3, '4': 7},
    '2': {'1': 8, '3': 2},
    '3': {'1': 5, '4': 1},
    '4': {'1': 2}
}

shortest_paths = floyd_warshall(fw_vertices, fw_graph)

print("Shortest distance matrix between all pairs:")
# Format output as a table
header = "   " + "  ".join(fw_vertices)
print(header)
print("-" * len(header) * 2)
for u in fw_vertices:
    row = [f"{shortest_paths[u][v]:2}" if shortest_paths[u][v] != math.inf else " ∞" for v in fw_vertices]
    print(f"{u} | " + " ".join(row))
```
